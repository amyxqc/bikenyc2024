```python
import pandas as pd
import glob
from tqdm import tqdm

# Define the path to your files
path = '/Users/amyxqc/Desktop/amy_bikenycfall2024/Data/CitiBike/*.csv'

# Initialize an empty DataFrame for the results
monthly_active_stations = pd.DataFrame()

# Process each file
for file in tqdm(sorted(glob.glob(path)), desc="Processing Files"):
    print(f"Processing file: {file}")
    df = pd.read_csv(file)
    df = df.drop_duplicates()

    # Ensure the required columns are present
    required_columns = ['started_at', 'start_station_id', 'end_station_id']
    if not all(col in df.columns for col in required_columns):
        print(f"Missing columns in file {file}. Skipping...")
        continue

    # Convert 'started_at' to datetime
    df['started_at'] = pd.to_datetime(df['started_at'], errors='coerce')
    df = df.dropna(subset=['started_at'])  # Drop rows with invalid dates

    # Extract year and month
    df['year'] = df['started_at'].dt.year
    df['month'] = df['started_at'].dt.month

    # Combine start and end stations into one DataFrame
    start_stations = df[['start_station_id', 'year', 'month']].rename(columns={'start_station_id': 'station_id'})
    end_stations = df[['end_station_id', 'year', 'month']].rename(columns={'end_station_id': 'station_id'})

    # Concatenate start and end stations and drop duplicates
    all_stations = pd.concat([start_stations, end_stations]).drop_duplicates()

    # Count unique station IDs per month (NYC total)
    monthly_counts = all_stations.groupby(['year', 'month'])['station_id'].nunique().reset_index(name='unique_active_stations')

    # Append results to the final DataFrame
    monthly_active_stations = pd.concat([monthly_active_stations, monthly_counts], ignore_index=True)

# Aggregate results across all files (if needed)
monthly_active_stations = monthly_active_stations.groupby(['year', 'month'])['unique_active_stations'].sum().reset_index()

# Display the results
print(monthly_active_stations)

# Save to CSV
output_path = '/Users/amyxqc/Desktop/amy_bikenycfall2024/Data/monthly_active_stations_nyc.csv'
monthly_active_stations.to_csv(output_path, index=False)
print(f"Monthly active stations data saved to {output_path}")
```
