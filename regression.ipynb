```python
import pandas as pd

# Define the file path
file_path = "/Users/amyxqc/Desktop/amy_bikenycfall2024/Data/final_combined_data_with_weather.csv"

# Load the CSV file into a Pandas DataFrame
combined_data = pd.read_csv(file_path)

# Display the first few rows of the DataFrame
print(combined_data)
```

```python
# Group by year and week, aggregating relevant numeric columns
nyc_total_data = combined_data.groupby(['year', 'week_number']).agg({
    'total_trips': 'sum',          # Sum total trips across boroughs
    'weekday_trips': 'sum',        # Sum weekday trips across boroughs
    'weekend_trips': 'sum',        # Sum weekend trips across boroughs
    'avg_daily_trips': 'mean',     # Average daily trips across boroughs
    'avg_weekday_trips': 'mean',   # Average weekday trips across boroughs
    'avg_weekend_trips': 'mean',   # Average weekend trips across boroughs
    'avg_precipitation': 'mean',   # Average precipitation across boroughs
    'avg_snow': 'mean',            # Average snow depth across boroughs
    'avg_temperature': 'mean',     # Average temperature across boroughs
    'avg_wind_speed': 'mean',      # Average wind speed across boroughs
}).reset_index()

# Display the first few rows of the new DataFrame
print(nyc_total_data)
```

```python
# Optionally save the new DataFrame to a CSV file
output_file_path = "/Users/amyxqc/Desktop/amy_bikenycfall2024/Data/nyc_total_data.csv"
nyc_total_data.to_csv(output_file_path, index=False)
print(f"NYC total data saved to {output_file_path}")
```

```python
# Assuming `combined_data` is the DataFrame you have displayed in the screenshot
# Calculate min, max, mean, and std for avg_daily_trips
nyc_avg_daily_trips_stats = combined_data.groupby(['year', 'week_number']).agg(
    avg_daily_trips_min=('avg_daily_trips', 'min'),
    avg_daily_trips_max=('avg_daily_trips', 'max'),
    avg_daily_trips_mean=('avg_daily_trips', 'mean'),
    avg_daily_trips_std=('avg_daily_trips', 'std')
).reset_index()

# Replace NaN in std with 0 for groups with only one value
nyc_avg_daily_trips_stats['avg_daily_trips_std'] = nyc_avg_daily_trips_stats['avg_daily_trips_std'].fillna(0)

# Display the new DataFrame
print(nyc_avg_daily_trips_stats)
```

```python
# Calculate overall min, max, mean, and std for avg_daily_trips
overall_stats = combined_data.agg({
    'avg_daily_trips': ['min', 'max', 'mean', 'std']
}).transpose()

# Rename columns for clarity
overall_stats.columns = ['min', 'max', 'mean', 'std']
overall_stats.reset_index(inplace=True)
overall_stats.rename(columns={'index': 'Metric'}, inplace=True)

# Display the new DataFrame
print("Overall Statistics for Avg Daily Trips Across All Weeks:")
print(overall_stats)
```
